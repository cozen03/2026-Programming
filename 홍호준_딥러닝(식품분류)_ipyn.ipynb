```python
import gdown

url = "https://drive.google.com/file/d/1cyp2lWOgtV8Va5ujfWY4NNmV-H3pbu0I/view"

gdown.download(
    url=url,
    output='/content/data.csv',
    fuzzy=True
)
```

```python
import pandas as pd

df = pd.read_csv("data.csv")
df.head()
```

```python
import seaborn as sns
from matplotlib import pyplot as plt

sns.scatterplot(data = df,
                x = "sweet",
                y = "crunchy",
                hue = "class")
plt.show()
```

```python
X = df[['sweet', 'crunchy']]
y = df['class']
```

```python
from tensorflow.keras.models import Sequential #<-모델 초기화
from tensorflow.keras.layers import Dense #<-레이어 생성

model = Sequential() #<-모델 생성

model.add(Dense(32, input_dim=2, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(3, activation='softmax'))
```

```python
model.compile(loss='sparse_categorical_crossentropy',
                    optimizer='adam',
                    metrics=['accuracy'])
```

```python
model.fit(X, y, epochs=1000)
```

```python
loss, accuracy = model.evaluate(X, y)
```

```python
model.predict(X)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions

# 1. Keras 모델의 예측 결과를 클래스 인덱스(1차원)로 변환해 주는 래퍼 클래스 정의
class KerasClassifierWrapper:
    def __init__(self, model):
        self.model = model

    def predict(self, X):
        # softmax 확률값(N, 3) 중 가장 큰 값의 인덱스(N,)를 반환
        return np.argmax(self.model.predict(X, verbose=1), axis=1)

# 2. 데이터 준비
X_train = df[['sweet', 'crunchy']].values
y_train = df['class'].astype('category').cat.codes.values

# 3. 래퍼로 감싼 모델을 clf에 전달
wrapped_model = KerasClassifierWrapper(model)

plt.figure(figsize=(8, 6))
plot_decision_regions(X_train, y_train, clf=wrapped_model, legend=2)

plt.xlabel('sweet')
plt.ylabel('crunchy')
plt.title('Deep Learning Decision Boundary')
plt.show()
```
