### 2.3.1 간단한 구현부터

```python
def AND(x1, x2):
    w1, w2 = 0.5, 0.5
    theta = 0.7
    tmp = x1*w1 + x2*w2
    if tmp <= theta:
        return 0
    else:
        return 1
```

### 2.3.3 가중치와 편향 구현하기

***ch02/and_gate.py***

```python
import numpy as np

def AND(x1, x2):
    x = np.array([x1, x2]) #입력
    w = np.array([0.5, 0.5]) #가중치
    b = -0.7 #편향
    tmp = np.sum(w*x) + b

    if tmp <= 0:
        return 0
    else:
        return 1
```

```python
AND(1, 1)  # 1을 출력
```

***ch02/nand_gate.py***

```python
def NAND(x1, x2):
    x = np.array([x1, x2]) #입력
    w = np.array([-0.5, -0.5]) #가중치
    b = 0.7 #편향
    tmp = np.sum(w*x) + b

    if tmp <= 0:
        return 0
    else:
        return 1
```

```python
NAND(0, 0)  # 0을 출력
```

***ch02/or_gate.py***

```python
def OR(x1, x2):
    x = np.array([x1, x2]) #입력
    w = np.array([-0.1, -0.1]) #가중치
    b = 1 #편향
    tmp = np.sum(w*x) + b

    if tmp <= 0:
        return 0
    else:
        return 1
```

```python
OR(0, 0)
```
