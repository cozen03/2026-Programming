# 1.4 매개변수와 인수, 리턴값
---
우리는 함수를 호출할 때 **입력값(인수)**를 전달하고, 함수의 실행이 끝난 뒤에는 **결과값(리턴값)**을 전달받을 수 있습니다.
함수 정의 시 입력값을 받을 변수를 **매개변수(parameter)**라고 하며,
함수를 실제로 호출할 때 전달하는 값을 **인수(argument)**라고 합니다.
함수가 실행을 마친 뒤 돌려주는 값은 **리턴값(return)**이라 부릅니다.


```
def add(a, b): #a, b는 매개변수
  retrun a+b   #a+b는 리턴값

print(add(3,4)) # 3,4는 인수
```

# 1.6 여러개의 리턴값을 받을 수 있을까??
결과적으로 말하자면 파이썬에서 함수의 리턴값은 하나만 존재합니다. 즉, '**return**'문을 여러개 작성하더라도 첫 번째 '**retrun**'문을 만나면 함수의 실행이 종료되고 해당 값이 반환됩니다. 나머지'**return**'문은 실행되지 않습니다.    

**예시 코드를 살펴봅시다.**

````python
def sum_mul(a,b):
  return a+b
  return a*b
print(sum_mul(10,5))
````

앞의 예시코드에서 알 수 있듯이 함수의 리턴값이 하나만 반환되며, 여러 개의 값을 반환하려면 이후에 배울 튜플이나 리스트 등의 자료구조를 사용하여 하나의 값으로 반환해야 합니다.

````python
#1부터 입력값까지의 정수를 배열에 저장하여 반환하는 함수
def list_append(a):
  x=list()
  for i in range(1,a+1):
    x.append(i)
  return x
t_list = list_append(10)
print(t_list)
````

# 1.7 파이썬 지역변수와 전역변수

파이썬에서 변수는 전역 변수(global variable)와 지역 변수(local variable)로 나뉩니다.

## 전역 변수

전역 변수는 스크립트 전체에서 접근 가능한 변수입니다. 즉, 어디서든 참조할 수 있습니다.





````python
#전역변수를 활용한 함수
x = 10

def print_global():
    print(x)

print_global()
````

위의 예시에서 '***x***'는 전역 변수로, 함수 '***print_global()***' 내부에서도 사용할 수 있습니다.

## 지역 변수
지역 변수는 특정 범위 내에서만 유효한 변수입니다. 일반적으로 함수 내부에서 정의되며, **해당 함수 내부에서만 접근 가능합니다.** 함수가 호출될 때 생성되고, 함수가 종료될 때 사라집니다.

````python
#지역변수를 활용한 함수
def print_local():
    y = 20
    print(y)

print_local()  # 출력 결과: 20
````

위의 예시에서 '***y***'는 '***print_local()***' 함수 내부에서만 유효한 지역 변수입니다.

````python
x = 10  # 전역변수

def sample():
    x = 5  # 지역변수
    print("함수 안:", x)

sample()
print("함수 밖:", x)
````

## 전역변수와 지역 변수의 우선순위
만약 전역 변수와 지역 변수가 동일한 이름을 가지고 있다면, 함수 내부에서 해당 변수를 참조할 때 지역 변수가 우선적으로 사용됩니다. 즉, 같은 이름의 전역 변수는 함수 내에서 재정의된 지역 변수에 의해 가려지게 됩니다.

````python
#지역변수와 전역변수를 활용한 프로그램
x = 10

def print_variable():
    x = 20
    print(x)

print_variable()
print(x)
````

위의 예시에서 '***print_variable()***' 함수 내부에서 '***x***'가 재정의되었으므로 지역 변수인 '***x***'가 사용되어 출력 결과는 20이 됩니다.

### 💡**TIP**　global 키워드
**global**키워드는 함수 안에서 전역변수를 수정하고자 할 때 사용하는 키워드입니다. 기본적으로 함수 내부에서 변수를 대입하면 지역변수로간주되는데,

 이를 전역변수로 취급하려면 global을 활용하여 선언하면 됩니다.

````python
x = 10

def update():
    global x  # 전역변수를 사용하겠다고 선언
    x = x + 1
    print(x)

update()     # 출력: 11
print(x)     # 출력: 11
````

##lamda함수(익명함수)
'**lambda**'는 익명함수를 생성할 때 사용하는 예약입니다. **lambda**'는 함수를 한 줄로 간결하게 만들때 사용하고 함수의 이름을 정의하지 않습니다. 만약 이름을 붙여 호출하고 싶다면 변수에 담아서 호출할 수 있습니다.
사용법은 다음과 같습니다.


```
#함수를 바로 사용할때
lambda 매개변수: 표현식

#변수에 담아서 사용할때
변수 = lambda 매개변수: 표현식
```


````python
#이름 없이 바로 사용하기
print((lambda x: x * x)(5))
````

````python
#변수에 담아서 사용하기
add = lambda a, b: a+b
result = add(3,4)
print(result)
````

### 💡**TIP**　삼항연산자
삼항 연산자는 조건에 따라 값을 선택하는 표현식입니다. Python에서 삼항 연산자는 if ~ else를 한 줄로 줄여 쓸 수 있는 조건식입니다. 사용법은 다음과 같습니다.
```
#<참일때 실행될명령어> if <조건식> else <거짓일때 실행될 명령어>

```

````python
age = 20
result = "성인" if age>=20 else "청소년"
print(result)
````

# 1.8 재귀함수
<font size=5><b>재귀함수</b>는 함수 안에 자신의 함수를 다시 호출하는 함수를 의미합니다. 이러한 재귀함수는 자신의 코드를 내부적으로 반복하다가, 일정한 조건이 만족되면 함수를 이탈하여 결과를 도출합니다.

재귀함수는 친구에게 도움을 요청하는 함수라고 생각하면 이해하기 쉽습니다.</font>

나보다 한 단계 더 쉬운 문제를 친구에게 부탁하고,
그 친구도 또다시 다른 친구에게 부탁하고…
결국 가장 쉬운 문제를 만난 친구가 먼저 해결하고,

````python
def factorial(n):
  if n==1:
    return 1
  else:
    return n*factorial(n-1)
print(factorial(4))
````

<font size =5>🧩 문제: 피보나치 수열의 n번째 항 구하기<br>
<font size =4>🔍 피보나치 수열(Fibonacci Sequence) 정의
**피보나치 수열**은 다음과 같이 정의됩니다:


*   첫 번째 항: F(0) = 0
*   두 번째 항: F(1) = 1
*   세 번째 항부터는 이전 두 항의 합: F(n) = F(n-1) + F(n-2)



<font size =5>다음 조건에 해당하는 프로그램을 작성하세요.


*  정수 n이 주어졌을 때, 피보나치 수열의 n번째 항을 재귀 함수를 이용해 구하세요.

　　　&nbsp;**입력**　　　　　　　　　　　　　　　**출력**
---
>fibonacci(6)　　　　　　　　　　　　　　　　　　　　　8
>fibonacci(7)　　　　　　　　　　　　　　　　　　　　　13




````python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
fibonacci(7)
````

<font size =5>🧩 문제: 문자열이 회문인지 판별하기<br>
<font size =4>🔍 회문(Palindrome) 정의

**회문(Palindrome)**이란,
문자열을 앞에서 읽든, 뒤에서 읽든 같은 순서가 되는 문자열을 말합니다.

*   예

  *   "level" → 앞에서 읽어도 level, 뒤에서 읽어도 level → ✅ 회문
  *   "madam" → 앞뒤가 같음 → ✅ 회문
  *   "hello" → hello ≠ olleh → ❌ 회문 아님

<font size =5>다음 조건에 해당하는 프로그램을 작성하세요.


*  입력된 문자열이 회문인지 판단하는 함수를 작성하세요
*  재귀함수를 사용하세요

　　　&nbsp;**입력**　　　　　　　　　　　　　　　**출력**
---
>is_palindrome("level")　　　　　　　　　　　　회문입니다.
>is_palindrome("hello")　　　　　　　　　　　　회문이 아닙니다.




````python
def is_palindrome(s):
    if len(s) <= 1:
        return True
    elif s[0] != s[-1]:
        return False
    elif s[0] == s[-1]:
        return is_palindrome(s[1:-1])
is_palindrome("hello")
````
